## 1. Đọc và khám phá tổng quan bộ dữ liệu (Loan Approval)

Bước này thực hiện nạp dữ liệu từ file `Loan_approval_data_2025.csv`, sau đó kiểm tra nhanh:
- **Kích thước tập dữ liệu:** Số dòng (mẫu) và số cột (thuộc tính).
- **Cấu trúc chi tiết từng cột:** Kiểu dữ liệu (`Dtype`), số lượng giá trị bị khuyết (`Null`), số giá trị duy nhất (`Unique`) cùng các giá trị min/max hoặc nhãn đại diện.
- **Xem trước dữ liệu:** Hiển thị 5 dòng đầu tiên để nắm cấu trúc thực tế của bảng.

## 2. [N] Nguồn và mô tả dữ liệu 
* **Nguồn dữ liệu:** Bộ dữ liệu **`Loan_approval_data_2025.csv`** (Realistic Loan Approval Dataset - Kaggle, cập nhật năm 2025).
* **Điều kiện sử dụng:** Dữ liệu mở (*Public/Open Data*) phục vụ mục đích học tập, nghiên cứu học máy và kiểm toán tính công bằng của thuật toán (*Algorithmic Fairness*), không chứa thông tin định danh cá nhân thật (PII đã được ẩn danh thành `CUST...`).
* **Quy mô dữ liệu:**
  * Số lượng mẫu (*Rows/Samples*): **50.000 bản ghi**.
  * Số lượng thuộc tính (*Columns*): **20 cột** (gồm 1 cột định danh `customer_id`, 18 cột đặc trưng đầu vào và 1 cột nhãn mục tiêu `loan_status`).
  * Phân loại kiểu dữ liệu: **3 biến phân loại kiểu chuỗi (`object`)**, **8 biến số nguyên (`int64`)** và **9 biến số thực (`float64`)**.

### Từ điển dữ liệu chi tiết (Data Dictionary)

| STT | Tên đặc trưng (Column) | Kiểu dữ liệu gốc | Vai trò / Nhóm biến | Đơn vị / Miền giá trị thực tế trong tập dữ liệu | Ý nghĩa nghiệp vụ chi tiết |
| :-: | :--- | :---: | :---: | :--- | :--- |
| 1 | `customer_id` | `object` | Định danh (Loại bỏ) | `CUST100000` $\rightarrow$ `CUST149999` | Mã số định danh duy nhất của mỗi hồ sơ khách hàng. |
| 2 | `age` | `int64` | Đầu vào (Số / Nhạy cảm) | $18 \rightarrow 70$ (Tuổi) | Độ tuổi của người nộp đơn vay vốn (dùng để kiểm toán tính công bằng). |
| 3 | `occupation_status` | `object` | Đầu vào (Phân loại) | `Employed`, `Self-Employed`, `Student` | Tình trạng nghề nghiệp hiện tại của khách hàng. |
| 4 | `years_employed` | `float64` | Đầu vào (Số liên tục) | $0.0 \rightarrow 39.9$ (Năm) | Số năm thâm niên làm việc trong nghề hiện tại. |
| 5 | `annual_income` | `int64` | Đầu vào (Số liên tục) | $15.000 \rightarrow 250.000$ (USD/năm) | Tổng thu nhập hàng năm của người vay. |
| 6 | `credit_score` | `int64` | Đầu vào (Số liên tục) | $348 \rightarrow 850$ (Điểm FICO) | Điểm xếp hạng tín dụng đánh giá mức độ uy tín trả nợ. |
| 7 | `credit_history_years` | `float64` | Đầu vào (Số liên tục) | $0.0 \rightarrow 30.0$ (Năm) | Thời gian khách hàng bắt đầu có lịch sử giao dịch tín dụng. |
| 8 | `savings_assets` | `int64` | Đầu vào (Số liên tục) | $0 \rightarrow 300.000$ (USD) | Tổng giá trị tiền gửi tiết kiệm và tài sản thanh khoản. |
| 9 | `current_debt` | `int64` | Đầu vào (Số liên tục) | $60 \rightarrow 163.344$ (USD) | Tổng dư nợ hiện tại khách hàng đang phải gánh trả. |
| 10 | `defaults_on_file` | `int64` | Đầu vào (Nhị phân) | $0$ (Không), $1$ (Có) | Ghi nhận khách hàng đã từng bị vỡ nợ/xù nợ trong quá khứ hay chưa. |
| 11 | `delinquencies_last_2yrs` | `int64` | Đầu vào (Số rời rạc) | $0 \rightarrow 9$ (Lần) | Số lần thanh toán nợ trễ hạn trong vòng 2 năm gần nhất. |
| 12 | `derogatory_marks` | `int64` | Đầu vào (Số rời rạc) | $0 \rightarrow 4$ (Lần) | Số điểm ghi nhận tiêu cực nghiêm trọng (bị kiện đòi nợ, tịch thu tài sản...). |
| 13 | `product_type` | `object` | Đầu vào (Phân loại) | `Credit Card`, `Personal Loan`, `Line of Credit` | Loại hình sản phẩm tín dụng mà khách hàng đăng ký. |
| 14 | `loan_intent` | `object` | Đầu vào (Phân loại) | `Business`, `Home Improvement`, `Debt Consolidation`, `Education`, `Personal`, `Medical` | Mục đích sử dụng khoản vay của khách hàng. |
| 15 | `loan_amount` | `int64` | Đầu vào (Số liên tục) | $500 \rightarrow 100.000$ (USD) | Số tiền gốc khách hàng đề nghị vay. |
| 16 | `interest_rate` | `float64` | Đầu vào (Số liên tục) | $6.0\% \rightarrow 23.0\%$ (%/năm) | Mức lãi suất áp dụng cho khoản vay được đề xuất. |
| 17 | `debt_to_income_ratio` | `float64` | Đầu vào (Số liên tục) | $0.002 \rightarrow 0.800$ (Tỷ lệ DTI) | Tỷ lệ tổng nợ hàng tháng trên thu nhập hàng tháng. |
| 18 | `loan_to_income_ratio` | `float64` | Đầu vào (Số liên tục) | $0.008 \rightarrow 2.001$ (Tỷ lệ LTI) | Tỷ lệ tổng số tiền vay trên thu nhập hàng năm ($\frac{\text{loan\_amount}}{\text{annual\_income}}$). |
| 19 | `payment_to_income_ratio`| `float64` | Đầu vào (Số liên tục) | $0.003 \rightarrow 0.667$ (Tỷ lệ PTI) | Tỷ lệ khoản trả góp định kỳ của khoản vay trên thu nhập. |
| 20 | **`loan_status`** | `int64` | **Đầu ra (Nhãn $y$)** | **$0$ (Từ chối - 44,95%), $1$ (Duyệt - 55,05%)** | **Quyết định phê duyệt khoản vay cuối cùng.** |

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

df = pd.read_csv('../data/Loan_approval_data_2025.csv')

print(f"1. Kích thước bộ dữ liệu: {df.shape[0]:,} mẫu (dòng) x {df.shape[1]} thuộc tính (cột)")
print("-" * 80)


summary_table = pd.DataFrame({
    'Kiểu dữ liệu (Dtype)': df.dtypes,
    'Số ô khuyết (Null)': df.isnull().sum(),
    'Số giá trị duy nhất (Unique)': df.nunique(),
    'Giá trị nhỏ nhất / Mẫu': [df[col].min() if df[col].dtype != 'object' else df[col].unique()[:3] for col in df.columns],
    'Giá trị lớn nhất / Mẫu': [df[col].max() if df[col].dtype != 'object' else f"Tổng {df[col].nunique()} nhãn" for col in df.columns]
})

print("2. Bảng kiểm tra nhanh cấu trúc 20 cột dữ liệu:")
display(summary_table)

print("-" * 80)
print("3. Xem 5 bản ghi đầu tiên của bộ dữ liệu:")
display(df.head())
```
