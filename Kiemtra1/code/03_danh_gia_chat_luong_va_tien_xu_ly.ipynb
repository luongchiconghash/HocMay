| :-: | :--- | :--- | :--- | :--- |
| **1** | **Bản ghi trùng lặp (Duplicates)** | Toàn bộ 20 cột | **0 bản ghi trùng (0,0%)** | Dữ liệu sạch về tính duy nhất của từng hồ sơ khách hàng. |
